```python
import json
from pathlib import Path

DIR_PATH = '../eval/results/missed_cases.json'
```

```python
with open(DIR_PATH, 'r+') as f:
    misses = json.load(f)
```

```python
import random

random.seed(42)
ca_to_tc = random.choices(misses['ca']['tc'], k=30)
```

```python
with open('../eval/results/sampled_ca_to_tc_miss.json', 'w', encoding='utf-8') as f:
    json.dump(ca_to_tc, f)
```
